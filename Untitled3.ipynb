# K-Fold Cross-Validation for Keras Sequential Model

This notebook demonstrates how to perform K-fold cross-validation on a Keras Sequential model. We will use the `sonar3.csv` dataset, which is a binary classification problem.

```python
# 1. Import necessary libraries
import os
import pandas as pd
from sklearn.model_selection import KFold
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.callbacks import ModelCheckpoint

print("Libraries imported successfully.")
```

## 2. Load and Prepare Data

We will load the `sonar3.csv` dataset. This dataset contains 60 features and a binary target variable (rock or mine).

```python
import pandas as pd # pandas 라이브러리 임포트 추가

# Load the dataset
df = pd.read_csv("https://raw.githubusercontent.com/SoongMoo/data/refs/heads/main/sonar3.csv") # 'sonar3.csv' 데이터셋: 음파 탐지기 신호를 기반으로 암석과 광물(Mine)을 구분하는 이진 분류 문제에 사용된 데이터셋입니다.

# Separate features (X) and target (y)
# X contains the first 60 columns (features)
X = df.iloc[:, 0:60]
# y contains the last column (target variable)
y = df.iloc[:, 60]

print(f"Dataset loaded. Shape: {df.shape}")
print("First 5 rows of features (X):\n", X.head())
print("First 5 rows of target (y):\n", y.head())
```

## 3. K-Fold Cross-Validation Setup

Define the number of folds and initialize the `KFold` object. We will use `shuffle=True` to randomize the data split for each fold.

```python
from sklearn.model_selection import KFold # KFold 클래스 임포트 추가

# Define the number of folds
k = 5

# Initialize KFold for splitting data into k folds
kfold = KFold(n_splits=k, shuffle=True, random_state=42) # K-Fold 교차 검증: 데이터를 K개의 폴드로 나누고, K번의 학습과 평가를 반복하여 모델 성능의 신뢰성을 높이는 방법입니다. # Added random_state for reproducibility

# List to store accuracy scores from each fold
acc_score = []

print(f"{k}-fold cross-validation setup complete.")
```

## 4. Define the Keras Model

We define a function `model_fn` that creates and returns our Keras Sequential model. This ensures a fresh model is built for each fold.

The model consists of an `Input` layer, two `Dense` hidden layers with 'relu' activation, and a final `Dense` output layer with 'sigmoid' activation for binary classification.

```python
def model_fn():
  model = Sequential() # Keras Sequential Model: 딥러닝 모델을 층(layer)을 순차적으로 쌓아 올리는 간단한 방식으로 구성하는 Keras API입니다.
  # Add an Input layer as the first layer to explicitly define the input shape
  # The sonar dataset has 60 features.
  model.add(Input(shape=(60,)))
  # First Dense hidden layer
  model.add(Dense(24, activation='relu'))
  # Second Dense hidden layer
  model.add(Dense(10, activation='relu'))
  # Output layer for binary classification (0 or 1)
  model.add(Dense(1, activation='sigmoid'))
  return model

print("Model function 'model_fn' defined.")
```

## 5. Perform K-Fold Cross-Validation

This loop iterates through each fold, splitting the data into training and test sets, building and compiling a new model, training it, and evaluating its performance. The best model from each fold (based on validation loss) is saved.

```python
# Prepare directory for saving models from each fold
model_save_dir = 'best_models'
os.makedirs(model_save_dir, exist_ok=True)

print(f"Models will be saved to: {model_save_dir}")

# Perform K-fold cross-validation
for fold, (train_index, test_index) in enumerate(kfold.split(X)):
  print(f"\n--- Training Fold {fold+1}/{k} ---")

  # Split data for the current fold
  X_train, X_test = X.iloc[train_index, :], X.iloc[test_index, :]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  # Create a new model for each fold to ensure independent training
  model = model_fn()

  # Compile the model for binary classification
  model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

  # Define ModelCheckpoint to save the best model from this fold
  # The model is saved if 'val_loss' improves. The filename will include the fold number and validation accuracy.
  filepath = os.path.join(model_save_dir, f"model_fold_{fold+1}_val_acc_{{val_accuracy:.4f}}.keras")
  checkpointer = ModelCheckpoint(filepath=filepath, monitor='val_loss', verbose=0, save_best_only=True) # ModelCheckpoint: 학습 중 가장 성능이 좋은 모델을 자동으로 저장하는 Keras 콜백 함수입니다.

  # Train the model with a validation split and callbacks
  # validation_split carves out a portion of the training data for validation during training
  # verbose=0 suppresses training output for cleaner logs.
  history = model.fit(X_train, y_train, epochs=50, batch_size=500,
                      validation_split=0.25, callbacks=[checkpointer], verbose=0)

  # Evaluate the model on the test set for this fold
  _, accuracy = model.evaluate(X_test, y_test, verbose=0)
  acc_score.append(accuracy)
  print(f"Fold {fold+1} Test Accuracy: {accuracy:.4f}")

print("\nK-fold cross-validation complete.")
```

## 6. Summarize Results

Calculate and print the average accuracy across all folds to get an overall performance estimate of the model.

```python
# Calculate and print the average accuracy across all folds
avg_acc_score = sum(acc_score) / k
print(f"\nAccuracy scores for each fold: {acc_score}")
print(f"Average K-fold Cross-Validation Accuracy: {avg_acc_score:.4f}")
```

```python
# Import necessary libraries for image processing and model explanation
import numpy as np # Import numpy for array manipulation (e.g., adding batch dimension)
from tensorflow.keras.applications.vgg16 import VGG16 # VGG16 모델: 대규모 이미지 분류 작업에 널리 사용되는 합성곱 신경망(CNN) 모델입니다. ImageNet 데이터셋으로 사전 학습되었습니다.
from tf_explain.core.grad_cam import GradCAM # tf_explain: 텐서플로우 모델의 설명을 돕는 라이브러리로, Grad-CAM과 같은 시각화 기법을 제공합니다.
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.applications.vgg16 import preprocess_input, decode_predictions # preprocess_input: VGG16 모델이 요구하는 형식(예: 픽셀 값 정규화, 채널 순서 변경)에 맞게 이미지를 전처리하는 함수입니다.

# Load the VGG16 model with pre-trained ImageNet weights
# This model will be used for feature extraction and classification
model = VGG16(weights="imagenet")

# Initialize the GradCAM explainer
# GradCAM helps visualize which parts of the input image are important for the model's prediction
explainer = GradCAM() # Grad-CAM (Gradient-weighted Class Activation Mapping): 딥러닝 모델이 특정 예측을 내릴 때, 이미지의 어느 영역에 집중했는지 시각화하여 보여주는 기술입니다.
```

```python
# Define a list of input image names (categories)
# These names will be used to construct image file paths.
input_list = ["maltese", "persian_cat", "squirrel_monkey", "grand_piano", "yawl"]
```

```python
# Define a list of ImageNet class indices corresponding to the input_list
# These indices are used by GradCAM to specify the target class for explanation.
imagenet_index = ["153", "283", "382", "579", "914"]
```

```python
# Loop through each image name and its corresponding ImageNet index
for l, i in zip(input_list, imagenet_index):
  # Load the image from the created dummy files, resizing it to 224x224 pixels
  img = load_img('./data/img/{}_0.jpg'.format(l), target_size=(224, 224))
  # Convert the PIL image instance to a NumPy array
  x = img_to_array(img)
  # Expand dimensions to create a batch of 1 image: (1, 224, 224, 3)
  # This is required as Keras models expect a batch dimension even for a single image.
  x = np.expand_dims(x, axis=0)
  # Preprocess the image for the VGG16 model (e.g., mean subtraction, scaling).
  # This ensures the input format matches what the pre-trained VGG16 model expects.
  x = preprocess_input(x) # preprocess_input: VGG16 모델이 요구하는 형식(예: 픽셀 값 정규화, 채널 순서 변경)에 맞게 이미지를 전처리하는 함수입니다.

  # Prepare the data in the format expected by GradCAM explainer: (input_images, labels)
  # Here, input_images is a batch containing our preprocessed image. Labels are not needed for explanation generation, so None.
  data = (x, None)

  # Generate the GradCAM heatmap for the specified input, model, target class, and layer
  # `int(i)` converts the string ImageNet class index to an integer, which is the target class for explanation.
  # `layer_name="block5_conv3"` specifies the convolutional layer from which to extract features for the heatmap.
  # This layer is often chosen as it's typically the last convolutional layer before classification, capturing high-level features.
  grid = explainer.explain(data, model, int(i), layer_name="block5_conv3") # Grad-CAM (Gradient-weighted Class Activation Mapping): 딥러닝 모델이 특정 예측을 내릴 때, 이미지의 어느 영역에 집중했는지 시각화하여 보여주는 기술입니다. # block5_conv3: VGG16 모델의 특정 층(Layer) 이름으로, Grad-CAM 시각화 시 어떤 층의 활성화 맵을 기반으로 설명할지 지정하는 데 사용됩니다.

  # Save the generated GradCAM heatmap as an image file.
  # The output 'grid' is a NumPy array representing the heatmap, which needs to be saved to disk for later visualization.
  # The filename includes the image label to easily identify the corresponding GradCAM result.
  explainer.save(grid, "./data/img", f"{l}_gradcam.jpg")
  print(f"Generated and saved GradCAM for {l} (class {i}) to ./data/img/{l}_gradcam.jpg")
```

```python
import os
from PIL import Image

# Define the directory where dummy images will be stored
output_dir = './data/img'
# Create the directory if it does not already exist
os.makedirs(output_dir, exist_ok=True)

# Loop through each item in the input_list to create a corresponding dummy image
for l in input_list:
    # Construct the full path for the dummy image file
    dummy_image_path = os.path.join(output_dir, f"{l}_0.jpg")
    # Check if the dummy image already exists to avoid overwriting
    if not os.path.exists(dummy_image_path):
        # Create a blank white RGB image of size 224x224 pixels
        # This size is commonly used as input for VGG16
        img = Image.new('RGB', (224, 224), color='white')
        # Save the blank image as a JPEG file
        img.save(dummy_image_path, "JPEG")
        print(f"Created dummy image: {dummy_image_path}")
    else:
        print(f"Dummy image already exists: {dummy_image_path}")

print("Dummy images setup complete. You can now proceed to run the cell that loads and processes these images.")
```

```python
import glob
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Lists to store original and Grad-CAM images for visualization
images_originals = []
images_cams = []

# Populate images_originals with original dummy images
# Loop through all files ending with '_0.jpg' in the './data/img/' directory
for img_path in glob.glob('./data/img/*_0.jpg'):
    # Read each image using matplotlib's image reader and append to the list
    images_originals.append(mpimg.imread(img_path))

# Populate images_cams with generated Grad-CAM images
# Loop through all files ending with '_gradcam.jpg' in the './data/img/' directory
for img_path in glob.glob('./data/img/*_gradcam.jpg'):
    # Read each Grad-CAM image and append to the list
    images_cams.append(mpimg.imread(img_path))

# Determine the number of images to display (assuming same count for originals and CAMS)
num_images = len(images_originals)

# Create a figure to display both original and Grad-CAM images side-by-side
# The figure size is adjusted for clarity, showing two columns (original, Grad-CAM)
# and 'num_images' rows.
plt.figure(figsize=(10, 5 * num_images)) # Adjust figure size dynamically

# Loop to display original and Grad-CAM images
for i in range(num_images):
    # Display original image in the first column
    plt.subplot(num_images, 2, 2 * i + 1) # (rows, columns, current_plot_index)
    plt.imshow(images_originals[i])
    plt.title(f"Original: {input_list[i]}") # Add title from input_list
    plt.axis('off') # Turn off axis for cleaner look

    # Display Grad-CAM image in the second column
    plt.subplot(num_images, 2, 2 * i + 2)
    plt.imshow(images_cams[i])
    plt.title(f"Grad-CAM: {input_list[i]}") # Add title from input_list
    plt.axis('off') # Turn off axis for cleaner look

plt.tight_layout() # Adjust subplot parameters for a tight layout
plt.show() # Display the plots
```

```python
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.applications import VGG16

# XAI알고리즘을 불러오는 부분입니다.
from tf_explain.core.grad_cam import GradCAM
from tf_explain.core.occlusion_sensitivity import OcclusionSensitivity

import glob
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# 원본 이미지가 들어갈 리스트 만들기
images_originals = []

# 원본 이미지가 저장된 폴더에서 하나씩 불러와 리스트에 넣기
for img_path in glob.glob('./data/img/*_0.jpg'):
    images_originals.append(mpimg.imread(img_path))

# 주피터 노트북 상에서 보여질 이미지의 크기
plt.figure(figsize=(20,20))

# 원본 이미지를 주피터 노트북에서 보이게 하기
for i, image_o in enumerate(images_originals):
    plt.subplot(5, 5, i + 1) # plt.subplot(nrows, ncols, index) nrows:세로로 나열되는 플롯 수, ncols:로로 나열되는 플롯 수, index:현재 위치한 플롯의 인덱스
    plt.imshow(image_o)
```

```python
# 사전에 학습된 딥러닝 모델 불러오기
model = VGG16(weights="imagenet", include_top=True)

# 원본 이미지 이름과 Imagenet에서의 해당 이미지 인덱스
input_list = ["maltese", "persian_cat", "squirrel_monkey", "grand_piano", "yawl"]
imagenet_index = ["153", "283", "382", "579", "914"]


# 여기서부터가 우리하려는 딥러닝부분이다.
# 그레이디언트 CAM 알고리즘 선택
explainer = GradCAM()    # 그레이디언트 CAM알고리즘 선택

# 그레이디언트 CAM 알고리즘이 적용된 이미지가 들어갈 빈 리스트 만들기
images_cams = []

# 그레이디언트 CAM 알고리즘 실행
# 가지고 있는 이미지를 불러온다.
for l, i in zip(input_list, imagenet_index):
    img = load_img('./data/img/{}_0.jpg'.format(l), target_size=(224, 224)) # 이미지를 불러오고 내부에서 처리될 이미지의 크기를 설정합니다.
    img = img_to_array(img) # 이미지를 넘파이 배열로 바꾸어 줍니다.
    data = ([img], None)
# 이미지 불러오기 끝
                                        # int(i) : index로 이미지를 불러와 model과 비교하도록 만들어 준다,
                                        #          data와 model을 비교할 때 사용되는 class
 # output=explainer.explain(데이터, 모델, 클래스)
    grid = explainer.explain(data, model, int(i), layer_name="block5_conv3") # 그레이디언트 CAM이 실행되는 부분입니다.
                                                  # layer_name="block5_conv3는 생략 가능
                                                  # block5_conv3는 CNN 모델, 특히 VGG 모델에서 5번째 블록의 3번째 합성곱 레이어를 사용한다는 의미
    explainer.save(grid, ".", './data/img/{}_cam.jpg'.format(l))# 실행 후 저장되는 이름입니다.
                                        # explainer.save(output, 저장폴더, 저장될 이름)

# 그레이디언트 CAM 알고리즘이 적용된 이미지를 불러오는 부분의 시작입니다.
plt.figure(figsize=(20,20))

for img_path in glob.glob('./data/img/*_cam.jpg'):
    images_cams.append(mpimg.imread(img_path))

for i, image_c in enumerate(images_cams):
    plt.subplot(5, 5, i + 1)
    plt.imshow(image_c)
```

| **Keras Sequential Model** | 딥러닝 모델을 만들 때 사용하는 케라스(Keras) 도구의 한 종류입니다. 마치 레고 블록을 한 줄로 쌓듯이, 인공지능의 여러 계산 단계를 순서대로 연결하여 모델을 구성하는 간단한 방식입니다.                         |

| **`ModelCheckpoint`**  | 인공지능 모델을 훈련하는 동안, 가장 좋은 성능을 보였을 때의 모델 상태를 자동으로 저장해주는 기능입니다. 나중에 모델의 성능이 떨어지더라도, 가장 잘 작동했던 순간의 모델을 다시 불러와 사용할 수 있도록 해줍니다. |

### 2. Grad-CAM을 이용한 이미지 분류 모델 설명

| 용어                   | 쉬운 설명                                                                                                                                                                                             |
| :--------------------- | :---------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Grad-CAM (Gradient-weighted Class Activation Mapping)** | 인공지능 모델이 '이 사진은 고양이 사진이야!'라고 판단했을 때, 그 판단을 내리기 위해 사진의 어느 부분을 가장 중요하게 바라보았는지 열지도(heatmap) 형태로 시각화해서 보여주는 기술입니다. 모델의 '생각'을 우리가 눈으로 볼 수 있게 해줍니다. |
| **VGG16 모델**         | 아주 복잡한 이미지 분류 문제를 풀 때 많이 사용되는 강력한 인공지능 모델 중 하나입니다. 'ImageNet'이라는 수많은 이미지 데이터를 미리 공부해서, 어떤 그림을 보더라도 잘 이해하고 분류할 수 있도록 훈련되어 있습니다.      |
| **`tf_explain`**       | 텐서플로우(TensorFlow)라는 딥러닝 프로그램을 사용할 때, 모델이 왜 그런 결정을 내렸는지 이해하기 어렵잖아요? `tf_explain`은 Grad-CAM처럼 모델의 결정을 쉽게 설명해주고 시각적으로 보여주는 다양한 도구들을 모아놓은 라이브러리입니다. |
| **`preprocess_input`** | VGG16 모델이 이미지를 가장 잘 이해하고 처리할 수 있도록, 사진을 특정 방식(예: 픽셀 값의 범위 조절, 색상 채널 순서 변경 등)에 맞춰 미리 준비하는 작업입니다. 모델이 훈련되었던 방식과 동일하게 맞춰주는 과정입니다. |
| **`block5_conv3`**     | VGG16 모델은 여러 단계의 복잡한 계산 층(Layer)으로 이루어져 있습니다. `block5_conv3`는 그 층들 중 하나를 가리키는 이름으로, Grad-CAM으로 모델의 '생각'을 시각화할 때 어떤 층의 정보를 바탕으로 분석할지 지정하는 부분입니다. 주로 모델이 고차원적인 특징을 포착하는 마지막 합성곱 층이 선택됩니다. |

### 3. Occlusion Sensitivity를 이용한 이미지 분류 모델 설명

| 용어                   | 쉬운 설명                                                                                                                                                                                             |
| :--------------------- | :---------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Occlusion Sensitivity** | 인공지능 모델이 특정 이미지를 보고 판단할 때, 이미지의 어떤 부분이 그 판단에 가장 큰 영향을 미치는지 알아내는 방법입니다. 이미지의 일부 영역을 가려보면서(패치), 모델의 예측이 어떻게 변하는지를 관찰하여 중요도를 파악합니다. 예를 들어, 고양이 사진에서 눈을 가렸더니 '고양이'라는 예측이 크게 바뀌면, 눈이 중요한 부분이라는 것을 알 수 있죠. |
| **`patch_size`**       | `Occlusion Sensitivity`를 사용할 때, 이미지에서 가릴(차단할) 영역의 크기를 결정하는 값입니다. 이 `patch_size`가 클수록 한 번에 더 넓은 영역을 가리게 되고, 작을수록 더 세밀하게 이미지의 중요 부분을 탐색할 수 있습니다. |

이 노트북은 `sonar3.csv` 데이터셋을 사용하여 Keras 순차 모델의 K-Fold 교차 검증을 수행하고, `tf_explain` 라이브러리의 Grad-CAM 및 Occlusion Sensitivity를 활용하여 VGG16 모델이 이미지 분류를 위해 이미지의 어떤 부분에 집중하는지 시각화하며 모델의 결정을 설명했습니다.

```python
# fine_tune_model.py (추가 학습 모델)
# 필요한 딥러닝 관련 도구들을 불러옵니다.
from tensorflow.keras.models import load_model # 저장된 딥러닝 모델을 불러올 때 사용합니다.
from tensorflow.keras.preprocessing.sequence import pad_sequences # 길이가 다른 문장들을 같은 길이로 맞춰주는 도구입니다.
from tensorflow.keras.utils import to_categorical # 숫자를 원-핫 인코딩 방식으로 바꿔주는 도구입니다.
import pickle # 파이썬 객체를 파일로 저장하고 불러올 때 사용합니다.

# 이전에 학습하고 저장했던 번역 모델을 "translation_model.h5" 파일에서 불러옵니다.
model = load_model("translation_model.h5")
# 모델을 다시 컴파일하여 옵티마이저를 재설정합니다.
# 이전에 학습된 모델을 불러왔을 때는 옵티마이저 정보가 포함되지 않을 수 있으므로, 새로운 학습을 위해 다시 컴파일해야 합니다.
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy']) # 모델 학습 방식을 다시 설정합니다.

# 저장했던 한글 토크나이저를 불러옵니다. (한글 문장을 숫자로 바꿔줄 도구)
with open("tokenizer_ko.pkl", "rb") as f:
    tokenizer_ko = pickle.load(f)
# 저장했던 영어 토크나이저를 불러옵니다. (영어 문장을 숫자로 바꿔줄 도구)
with open("tokenizer_en.pkl", "rb") as f:
    tokenizer_en = pickle.load(f)
# 저장했던 모델의 설정 값들(최대 길이, 단어 사전 크기 등)을 불러옵니다.
with open("params.pkl", "rb") as f:
    max_encoder_len, max_decoder_len, Y_vocab_size = pickle.load(f)

# 사용자 입력 (새롭게 학습시킬 한글 문장과 그 번역을 입력받습니다)
user_ko = input("한글 문장 입력: ")
user_en = input("영어 번역 입력: ")

# 사용자가 입력한 영어 번역 문장에 '<start>' 토큰을 앞에 붙입니다. (모델 입력용)
user_en_input = "<start> " + user_en
# 사용자가 입력한 영어 번역 문장에 '<end>' 토큰을 뒤에 붙입니다. (모델 정답용)
user_en_target = user_en + " <end>"

# 입력받은 한글 문장을 숫자로 이루어진 시퀀스로 바꿉니다.
X_seq = tokenizer_ko.texts_to_sequences([user_ko])
# 입력받은 영어 입력 문장(`user_en_input`)을 숫자로 이루어진 시퀀스로 바꿉니다.
Y_input_seq = tokenizer_en.texts_to_sequences([user_en_input])
# 입력받은 영어 정답 문장(`user_en_target`)을 숫자로 이루어진 시퀀스로 바꿉니다.
Y_target_seq = tokenizer_en.texts_to_sequences([user_en_target])

# 한글 시퀀스의 길이를 인코더의 최대 길이에 맞춰 통일합니다.
X_seq = pad_sequences(X_seq, maxlen=max_encoder_len, padding='post')
# 영어 입력 시퀀스의 길이를 디코더의 최대 길이에 맞춰 통일합니다.
Y_input_seq = pad_sequences(Y_input_seq, maxlen=max_decoder_len, padding='post')
# 영어 정답 시퀀스의 길이를 디코더의 최대 길이에 맞춰 통일합니다.
Y_target_seq = pad_sequences(Y_target_seq, maxlen=max_decoder_len, padding='post')

# 영어 정답 시퀀스를 원-핫 인코딩 형태로 바꿉니다. (모델이 이해하기 쉬운 형태로 변환)
Y_target_oh = to_categorical(Y_target_seq, num_classes=Y_vocab_size)

# 추가 학습 (기존 모델에 새로운 데이터를 추가하여 학습시킵니다)
model.fit([X_seq, Y_input_seq], Y_target_oh, epochs=10, verbose=1)

# 변경된 모델을 "translation_model.h5" 파일로 다시 저장합니다.
model.save("translation_model.h5")
print("✅ 사용자 입력으로 추가 학습 및 저장 완료")
```

```python
# inference_fixed.py (번역 모델 사용하기)
# 딥러닝 모델을 불러오고, 모델의 특정 부분을 가져오는 데 필요한 도구들을 불러옵니다.
from tensorflow.keras.models import load_model, Model
# 모델의 입력층을 정의하는 데 사용됩니다.
from tensorflow.keras.layers import Input
# 길이가 다른 문장들을 같은 길이로 맞춰주는 도구입니다.
from tensorflow.keras.preprocessing.sequence import pad_sequences
# 숫자 계산을 효율적으로 할 수 있게 도와주는 도구입니다.
import numpy as np
# 파이썬 객체(토크나이저, 설정 값)를 파일로 저장하고 불러올 때 사용합니다.
import pickle

# 이전에 학습하고 저장했던 번역 모델을 "translation_model.h5" 파일에서 불러옵니다.
model = load_model("translation_model.h5")
# 저장했던 한글 토크나이저를 불러옵니다. (한글 문장을 숫자로 바꿔줄 도구)
with open("tokenizer_ko.pkl", "rb") as f: tokenizer_ko = pickle.load(f)
# 저장했던 영어 토크나이저를 불러옵니다. (영어 문장을 숫자로 바꿔줄 도구)
with open("tokenizer_en.pkl", "rb") as f: tokenizer_en = pickle.load(f)
# 저장했던 모델의 설정 값들(최대 길이, 단어 사전 크기 등)을 불러옵니다.
with open("params.pkl", "rb") as f: max_encoder_len, max_decoder_len, Y_vocab_size = pickle.load(f)

# ---------------------
# Encoder 모델 생성 (번역할 문장을 이해하는 부분만 따로 떼어냅니다)
# ---------------------
# 전체 모델에서 인코더의 입력층을 가져옵니다.
encoder_inputs = model.input[0]            # encoder_input
# 인코더의 임베딩 층을 가져옵니다.
encoder_embedding = model.get_layer("encoder_embedding")(encoder_inputs)
# 인코더의 LSTM 층을 가져옵니다.
encoder_lstm = model.get_layer("encoder_lstm")
# 인코더 LSTM 층의 출력과 핵심 상태(state_h, state_c)를 계산합니다.
encoder_outputs, state_h, state_c = encoder_lstm(encoder_embedding)
# 번역할 문장의 핵심 정보를 뽑아내는 인코더 모델을 만듭니다. (입력: 한글 문장, 출력: 핵심 상태)
encoder_model = Model(encoder_inputs, [state_h, state_c])

# ---------------------
# Decoder 모델 생성 (핵심 정보를 바탕으로 번역된 문장을 만들어내는 부분만 따로 떼어냅니다)
# ---------------------
# 전체 모델에서 디코더의 입력층을 가져옵니다.
decoder_inputs = model.input[1]            # decoder_input
# 디코더의 임베딩 층을 가져옵니다.
decoder_embedding_layer = model.get_layer("decoder_embedding")
# 디코더의 LSTM 층을 가져옵니다.
decoder_lstm_layer = model.get_layer("decoder_lstm")
# 디코더의 마지막 출력층을 가져옵니다.
decoder_dense = model.get_layer("decoder_dense")

# 디코더의 초기 상태(인코더에서 넘어온 핵심 정보)를 받을 입력층을 정의합니다.
decoder_state_input_h = Input(shape=(128,))
decoder_state_input_c = Input(shape=(128,))
# 디코더 임베딩 층에 디코더 입력을 넣습니다.
dec_emb2 = decoder_embedding_layer(decoder_inputs)

# 디코더 LSTM 층을 실행하고, 새로운 출력과 다음 상태를 얻습니다. (이전 상태를 사용)
decoder_outputs, state_h2, state_c2 = decoder_lstm_layer(
    dec_emb2, initial_state=[decoder_state_input_h, decoder_state_input_c]
)
# 최종적으로 디코더 출력층을 통과시켜 단어들의 확률 분포를 얻습니다.
decoder_outputs = decoder_dense(decoder_outputs)
# 번역된 문장을 한 단어씩 예측하는 디코더 모델을 만듭니다.
# (입력: 영어 단어, 이전 상태 / 출력: 다음 단어 예측, 새로운 상태)
decoder_model = Model(
    [decoder_inputs, decoder_state_input_h, decoder_state_input_c],
    [decoder_outputs, state_h2, state_c2]
)

# ---------------------
# translate 함수 (실제로 번역을 수행하는 함수)
# ---------------------
# 한글 텍스트를 입력받아 영어로 번역하는 함수입니다.
def translate(text):
    # 입력된 한글 텍스트를 숫자로 이루어진 시퀀스로 바꿉니다.
    seq = tokenizer_ko.texts_to_sequences([text])
    # 시퀀스의 길이를 인코더의 최대 길이에 맞춰 통일합니다.
    seq = pad_sequences(seq, maxlen=max_encoder_len, padding='post')

    # 인코더 모델을 사용하여 한글 문장의 핵심 상태(context vector)를 추출합니다.
    states = encoder_model.predict(seq)  # batch, hidden
    # 디코더의 시작을 알리는 '<start>' 토큰을 준비합니다.
    target_seq = np.array([[tokenizer_en.word_index["<start>"]]])
    # 번역 결과를 저장할 빈 리스트를 만듭니다.
    result = []

    # 디코더의 최대 길이만큼 반복하여 한 단어씩 예측합니다.
    for _ in range(max_decoder_len):
        # 디코더 모델을 사용하여 다음 단어를 예측하고, 새로운 상태를 얻습니다.
        output_tokens, h, c = decoder_model.predict([target_seq] + states)
        # 예측된 단어들 중에서 가장 확률이 높은 단어의 인덱스를 찾습니다.
        token_idx = np.argmax(output_tokens[0, -1, :])
        # 인덱스에 해당하는 단어를 찾습니다. (사전에 없으면 빈 문자열)
        word = tokenizer_en.index_word.get(token_idx, "")
        # 예측된 단어가 '<end>' 토큰이면 번역을 중단합니다.
        if word == "<end>": break
        # 예측된 단어를 결과 리스트에 추가합니다.
        result.append(word)
        # 현재 예측된 단어를 다음 예측의 입력으로 사용하기 위해 업데이트합니다.
        target_seq = np.array([[token_idx]])
        # 디코더의 상태를 업데이트합니다.
        states = [h, c]

    # 번역된 단어들을 공백으로 이어붙여 최종 번역 문장을 만듭니다.
    return " ".join(result)

# ---------------------
# 테스트 (번역기 사용해보기)
# ---------------------
# 이 파일이 직접 실행될 때만 아래 코드가 동작하도록 합니다.
if __name__ == "__main__":
    # 'q'를 입력하기 전까지 계속 번역을 할 수 있도록 반복합니다.
    while True:
        # 사용자에게 한글 문장을 입력받습니다.
        txt = input("한글 입력(q 종료): ")
        # 'q'를 입력하면 반복을 멈춥니다.
        if txt.lower() == "q": break
        # 입력된 한글 문장을 번역하고 결과를 출력합니다.
        print("번역:", translate(txt))
```

```python
# 2. 토크나이저 (텍스트를 숫자로 바꾸는 도구)
# 한글 문장을 숫자로 바꿔줄 토크나이저를 만듭니다.
tokenizer_ko = Tokenizer()
# 한글 토크나이저에게 X_texts 문장들을 학습시켜서, 어떤 단어가 몇 번 나오는지 등을 파악하게 합니다.
tokenizer_ko.fit_on_texts(X_texts)
# 영어 문장을 숫자로 바꿔줄 토크나이저를 만듭니다. (filters=''는 특별한 기호들을 제거하지 않도록 설정)
tokenizer_en = Tokenizer(filters='')
# 영어 토크나이저에게 Y_input과 Y_target 문장들을 학습시킵니다.
tokenizer_en.fit_on_texts(Y_input + Y_target)

# 한글 문장들을 숫자로 이루어진 시퀀스(순서열)로 바꿉니다.
X_seq = tokenizer_ko.texts_to_sequences(X_texts)
# 영어 입력 문장들을 숫자로 이루어진 시퀀스로 바꿉니다.
Y_input_seq = tokenizer_en.texts_to_sequences(Y_input)
# 영어 정답 문장들을 숫자로 이루어진 시퀀스로 바꿉니다.
Y_target_seq = tokenizer_en.texts_to_sequences(Y_target)

# 인코더(한글 처리 부분)의 최대 문장 길이를 찾습니다. (가장 긴 한글 문장의 길이)
max_encoder_len = max(len(s) for s in X_seq)
# 디코더(영어 처리 부분)의 최대 문장 길이를 찾습니다. (가장 긴 영어 문장의 길이)
max_decoder_len = max(len(s) for s in Y_input_seq)

# 한글 시퀀스들의 길이를 max_encoder_len에 맞춰 통일합니다. (짧은 문장은 뒤에 0을 채워 넣습니다.)
X_seq = pad_sequences(X_seq, maxlen=max_encoder_len, padding='post')
# 영어 입력 시퀀스들의 길이를 max_decoder_len에 맞춰 통일합니다.
Y_input_seq = pad_sequences(Y_input_seq, maxlen=max_decoder_len, padding='post')
# 영어 정답 시퀀스들의 길이를 max_decoder_len에 맞춰 통일합니다.
Y_target_seq = pad_sequences(Y_target_seq, maxlen=max_decoder_len, padding='post')

# 한글 단어 사전의 크기 (총 단어 수)를 계산합니다.
X_vocab_size = len(tokenizer_ko.word_index) + 1
# 영어 단어 사전의 크기 (총 단어 수)를 계산합니다.
Y_vocab_size = len(tokenizer_en.word_index) + 1
# 영어 정답 시퀀스를 원-핫 인코딩 형태로 바꿉니다. (모델이 이해하기 쉬운 형태로 변환)
Y_target_oh = to_categorical(Y_target_seq, num_classes=Y_vocab_size)

# 3. Seq2Seq 모델 구성 (번역 모델의 설계도)
# 인코더 (한글 문장을 이해하는 부분)
# 인코더에 들어갈 입력층을 정의합니다. (한글 문장의 최대 길이만큼)
encoder_inputs = Input(shape=(max_encoder_len,), name="encoder_input")
# 입력된 한글 단어들을 컴퓨터가 이해할 수 있는 벡터(숫자 배열)로 변환합니다.
enc_emb = Embedding(X_vocab_size, 64, name="encoder_embedding")(encoder_inputs)
# 변환된 벡터를 LSTM 층에 넣어 한글 문장의 핵심 정보를 추출합니다. (state_h, state_c가 그 핵심 정보입니다.)
encoder_outputs, state_h, state_c = LSTM(128, return_state=True, name="encoder_lstm")(enc_emb)

# 디코더 (영어 문장을 생성하는 부분)
# 디코더에 들어갈 입력층을 정의합니다. (영어 문장의 최대 길이만큼)
decoder_inputs = Input(shape=(max_decoder_len,), name="decoder_input")
# 입력된 영어 단어들을 컴퓨터가 이해할 수 있는 벡터로 변환합니다.
dec_emb = Embedding(Y_vocab_size, 64, name="decoder_embedding")(decoder_inputs)
# LSTM 층을 만들고, 인코더에서 얻은 핵심 정보를 바탕으로 영어 문장을 만듭니다.
decoder_lstm = LSTM(128, return_sequences=True, return_state=True, name="decoder_lstm")
# 디코더 LSTM 층을 통해 영어 문장 예측 결과를 얻습니다.
decoder_outputs, _, _ = decoder_lstm(dec_emb, initial_state=[state_h, state_c])
# 예측된 단어들을 최종 영어 단어 사전의 크기만큼 확률 분포로 변환하는 층입니다.
decoder_dense = Dense(Y_vocab_size, activation='softmax', name="decoder_dense")
# 최종 디코더의 출력입니다.
decoder_outputs = decoder_dense(decoder_outputs)

# 전체 Seq2Seq 모델을 만듭니다. (한글 입력과 영어 입력이 들어가서 영어 출력이 나오는 모델)
model = Model([encoder_inputs, decoder_inputs], decoder_outputs)
# 모델을 학습할 준비를 합니다. (어떤 방식으로 최적화하고, 손실을 계산하며, 평가할지 설정)
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
# 모델을 학습시킵니다. (X_seq와 Y_input_seq를 입력으로 Y_target_oh를 정답으로 하여 200번 반복 학습)
model.fit([X_seq, Y_input_seq], Y_target_oh, batch_size=2, epochs=200)

# 학습된 모델을 "translation_model.h5" 파일로 저장합니다.
model.save("translation_model.h5")
# 한글 토크나이저를 "tokenizer_ko.pkl" 파일로 저장합니다.
with open("tokenizer_ko.pkl", "wb") as f: pickle.dump(tokenizer_ko, f)
# 영어 토크나이저를 "tokenizer_en.pkl" 파일로 저장합니다.
with open("tokenizer_en.pkl", "wb") as f: pickle.dump(tokenizer_en, f)
# 인코더/디코더의 최대 길이와 영어 단어 사전 크기 같은 중요한 설정 값들을 "params.pkl" 파일로 저장합니다.
with open("params.pkl", "wb") as f: pickle.dump((max_encoder_len, max_decoder_len, Y_vocab_size), f)


print("학습 완료, 모델과 토크나이저 저장 완료")
```
