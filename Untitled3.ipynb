# K-Fold Cross-Validation for Keras Sequential Model

This notebook demonstrates how to perform K-fold cross-validation on a Keras Sequential model. We will use the `sonar3.csv` dataset, which is a binary classification problem.

````python
# 1. Import necessary libraries
import os
import pandas as pd
from sklearn.model_selection import KFold
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.callbacks import ModelCheckpoint

print("Libraries imported successfully.")
````

## 2. Load and Prepare Data

We will load the `sonar3.csv` dataset. This dataset contains 60 features and a binary target variable (rock or mine).

````python
# Load the dataset
df = pd.read_csv("https://raw.githubusercontent.com/SoongMoo/data/refs/heads/main/sonar3.csv")

# Separate features (X) and target (y)
# X contains the first 60 columns (features)
X = df.iloc[:, 0:60]
# y contains the last column (target variable)
y = df.iloc[:, 60]

print(f"Dataset loaded. Shape: {df.shape}")
print("First 5 rows of features (X):\n", X.head())
print("First 5 rows of target (y):\n", y.head())
````

## 3. K-Fold Cross-Validation Setup

Define the number of folds and initialize the `KFold` object. We will use `shuffle=True` to randomize the data split for each fold.

````python
k = 5

kfold = KFold(n_splits=k, shuffle=True, random_state=42)

acc_score = []

print(f"{k}-fold cross-validation setup complete.")
````

## 4. Define the Keras Model

We define a function `model_fn` that creates and returns our Keras Sequential model. This ensures a fresh model is built for each fold.

The model consists of an `Input` layer, two `Dense` hidden layers with 'relu' activation, and a final `Dense` output layer with 'sigmoid' activation for binary classification.

````python
def model_fn():
  model = Sequential()
  # Add an Input layer as the first layer to explicitly define the input shape
  # The sonar dataset has 60 features.
  model.add(Input(shape=(60,)))
  # First Dense hidden layer
  model.add(Dense(24, activation='relu'))
  # Second Dense hidden layer
  model.add(Dense(10, activation='relu'))
  # Output layer for binary classification (0 or 1)
  model.add(Dense(1, activation='sigmoid'))
  return model

print("Model function 'model_fn' defined.")
````

## 5. Perform K-Fold Cross-Validation

This loop iterates through each fold, splitting the data into training and test sets, building and compiling a new model, training it, and evaluating its performance. The best model from each fold (based on validation loss) is saved.

````python
# Prepare directory for saving models from each fold
model_save_dir = 'best_models'
os.makedirs(model_save_dir, exist_ok=True)

print(f"Models will be saved to: {model_save_dir}")

# Perform K-fold cross-validation
for fold, (train_index, test_index) in enumerate(kfold.split(X)):
  print(f"\n--- Training Fold {fold+1}/{k} ---")

  # Split data for the current fold
  X_train, X_test = X.iloc[train_index, :], X.iloc[test_index, :]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  # Create a new model for each fold to ensure independent training
  model = model_fn()

  # Compile the model for binary classification
  model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

  # Define ModelCheckpoint to save the best model from this fold
  # The model is saved if 'val_loss' improves. The filename will include the fold number and validation accuracy.
  filepath = os.path.join(model_save_dir, f"model_fold_{fold+1}_val_acc_{{val_accuracy:.4f}}.keras")
  checkpointer = ModelCheckpoint(filepath=filepath, monitor='val_loss', verbose=0, save_best_only=True)

  # Train the model with a validation split and callbacks
  # validation_split carves out a portion of the training data for validation during training
  # verbose=0 suppresses training output for cleaner logs.
  history = model.fit(X_train, y_train, epochs=50, batch_size=500,
                      validation_split=0.25, callbacks=[checkpointer], verbose=0)

  # Evaluate the model on the test set for this fold
  _, accuracy = model.evaluate(X_test, y_test, verbose=0)
  acc_score.append(accuracy)
  print(f"Fold {fold+1} Test Accuracy: {accuracy:.4f}")

print("\nK-fold cross-validation complete.")
````

## 6. Summarize Results

Calculate and print the average accuracy across all folds to get an overall performance estimate of the model.

````python
# Calculate and print the average accuracy across all folds
avg_acc_score = sum(acc_score) / k
print(f"\nAccuracy scores for each fold: {acc_score}")
print(f"Average K-fold Cross-Validation Accuracy: {avg_acc_score:.4f}")
````

````python
# 이미지 처리 및 모델 설명(Grad-CAM)에 필요한 라이브러리들을 불러옵니다.
import numpy as np # 숫자 배열 계산을 위한 라이브러리 (예: 이미지 차원 추가)
from tensorflow.keras.applications.vgg16 import VGG16 # 이미지 분류 모델인 VGG16을 불러옵니다.
from tf_explain.core.grad_cam import GradCAM # Grad-CAM 기능을 위한 라이브러리를 불러옵니다.
from tensorflow.keras.preprocessing.image import load_img, img_to_array # 이미지를 로드하고 배열로 변환하는 함수를 불러옵니다.
from tensorflow.keras.applications.vgg16 import preprocess_input, decode_predictions # VGG16 모델에 맞게 이미지를 전처리하고 예측 결과를 해석하는 함수를 불러옵니다.

# ImageNet 데이터셋으로 미리 학습된 VGG16 모델을 불러옵니다.
# 이 모델은 이미지에서 특징을 추출하고 이미지를 분류하는 데 사용됩니다.
model = VGG16(weights="imagenet")

# GradCAM 설명기(explainer)를 초기화합니다.
# GradCAM은 모델이 예측을 할 때 이미지의 어느 부분을 중요하게 보았는지 시각화하여 보여줍니다.
explainer = GradCAM()
````

````python
# 분석할 이미지들의 이름 목록을 정의합니다.
# 이 이름들은 나중에 이미지 파일 경로를 만들 때 사용됩니다.
input_list = ["maltese", "persian_cat", "squirrel_monkey", "grand_piano", "yawl"]
````

````python
# ImageNet 클래스 인덱스 목록을 정의합니다. 위 input_list에 있는 이미지들과 일치합니다.
# 이 인덱스는 Grad-CAM이 어떤 클래스(예측 결과)에 대한 설명을 생성할지 지정할 때 사용됩니다.
imagenet_index = ["153", "283", "382", "579", "914"]
````

````python
# 각 이미지 이름과 해당 ImageNet 인덱스를 이용하여 반복 작업을 수행합니다.
for l, i in zip(input_list, imagenet_index):
  # 더미 파일에서 이미지를 로드하고, VGG16 모델 입력 크기인 224x224 픽셀로 크기를 조절합니다.
  img = load_img('./data/img/{}_0.jpg'.format(l), target_size=(224, 224))
  # PIL 이미지를 NumPy 배열로 변환합니다.
  x = img_to_array(img)
  # 모델은 보통 한 번에 여러 이미지를 처리하므로, 현재 단일 이미지에 배치 차원을 추가합니다: (1, 224, 224, 3) 형태로 만듭니다.
  x = np.expand_dims(x, axis=0)
  # VGG16 모델이 예상하는 형식에 맞게 이미지를 전처리합니다. (예: 평균값 빼기, 스케일링).
  # 이는 미리 학습된 VGG16 모델이 올바르게 작동하도록 합니다.
  x = preprocess_input(x)

  # GradCAM 설명기가 필요로 하는 데이터 형식 (입력 이미지, 라벨)으로 준비합니다.
  # 여기서는 입력 이미지 배치만 필요하고, 라벨은 설명 생성에 필수가 아니므로 None으로 둡니다.
  data = (x, None)

  # Grad-CAM 히트맵을 생성합니다.
  # int(i)는 ImageNet 클래스 인덱스를 정수로 변환하여, 이 클래스에 대한 설명을 생성하도록 지정합니다.
  # layer_name="block5_conv3"는 히트맵을 생성할 때 VGG16 모델의 특정 레이어(여기서는 마지막 컨볼루션 레이어)의 특징을 사용하도록 지정합니다.
  # 이 레이어는 보통 이미지의 고수준 특징을 잘 포착합니다.
  grid = explainer.explain(data, model, int(i), layer_name="block5_conv3")

  # 생성된 Grad-CAM 히트맵을 이미지 파일로 저장합니다.
  # 'grid'는 히트맵을 나타내는 NumPy 배열이며, 나중에 시각화를 위해 디스크에 저장해야 합니다.
  # 파일 이름에 이미지 라벨을 포함하여 어떤 이미지의 Grad-CAM 결과인지 쉽게 알 수 있도록 합니다.
  explainer.save(grid, "./data/img", f"{l}_gradcam.jpg")
  print(f"{l} (클래스 {i}) 에 대한 Grad-CAM이 생성되어 ./data/img/{l}_gradcam.jpg 에 저장되었습니다.")
````

````python
import os
from PIL import Image

# 더미 이미지(테스트용 빈 이미지)를 저장할 디렉토리를 정의합니다.
output_dir = './data/img'
# 디렉토리가 없으면 새로 만듭니다. 이미 있으면 넘어갑니다.
os.makedirs(output_dir, exist_ok=True)

# input_list의 각 항목에 대해 더미 이미지를 만듭니다.
for l in input_list:
    # 더미 이미지 파일의 전체 경로를 만듭니다. (예: ./data/img/maltese_0.jpg)
    dummy_image_path = os.path.join(output_dir, f"{l}_0.jpg")
    # 이미 해당 더미 이미지가 존재하면 다시 만들지 않고 건너뜁니다.
    if not os.path.exists(dummy_image_path):
        # 224x224 크기의 흰색 RGB 이미지를 만듭니다.
        # 이 크기는 VGG16과 같은 이미지 모델의 일반적인 입력 크기입니다.
        img = Image.new('RGB', (224, 224), color='white')
        # 만들어진 빈 이미지를 JPEG 파일로 저장합니다.
        img.save(dummy_image_path, "JPEG")
        print(f"더미 이미지 생성: {dummy_image_path}")
    else:
        print(f"더미 이미지가 이미 존재합니다: {dummy_image_path}")

print("더미 이미지 설정 완료. 이제 이 이미지를 로드하고 처리하는 코드를 실행할 수 있습니다.")
````

````python
# 딥러닝 모델을 '순차적(Sequential)'으로 쌓아 올리겠다고 선언합니다.
model = Sequential()

# 첫 번째 레이어는 '임베딩(Embedding)' 레이어입니다.
# 이 레이어는 각 단어 인덱스(word_size, 21개 종류)를 8차원의 '숫자 벡터'로 변환합니다.
# 'input_length=4'는 각 문장이 4개의 단어로 구성된다고 모델에게 알려주는 역할을 합니다.
model.add(Embedding(word_size, 8, input_length=4))

# 이 모델의 입력 형태가 (None, 4)라는 것을 명시적으로 알려줍니다.
# None은 '배치 크기'로, 한 번에 몇 개의 문장을 처리할지는 유동적이라는 뜻입니다.
# 4는 'input_length'와 같이 각 문장의 길이가 4개 단어라는 뜻입니다.
model.build((None, 4))

# 'Flatten' 레이어는 이전 레이어의 출력 형태를 1차원 배열로 펼쳐줍니다.
# 예를 들어 (배치 크기, 4, 8) 형태의 임베딩 출력을 (배치 크기, 32)로 만듭니다. (4 * 8 = 32)
model.add(Flatten())

# 마지막 'Dense' 레이어는 최종 예측을 담당합니다.
# '1'은 출력 뉴런의 개수가 1개라는 뜻으로, 긍정/부정 이진 분류를 위한 출력입니다.
# 'sigmoid' 활성화 함수는 0과 1 사이의 값으로 예측을 출력하여 확률처럼 해석할 수 있게 합니다.
model.add(Dense(1, activation='sigmoid'))

# 모델의 구조를 요약하여 출력합니다. 각 레이어의 이름, 출력 형태, 파라미터(학습 가능한 가중치) 개수를 보여줍니다.
model.summary()

# 모델을 '컴파일(Compile)'합니다. 이는 모델 학습 전에 필요한 설정들을 준비하는 과정입니다.
# - 'optimizer='adam'': 모델의 가중치를 어떻게 업데이트할지 결정하는 최적화 도구입니다.
# - 'loss='binary_crossentropy'': 이진 분류 문제에 적합한 손실 함수로, 모델의 예측이 얼마나 틀렸는지 측정합니다.
# - 'metrics=['accuracy']': 모델의 성능을 '정확도(accuracy)'로 평가하겠다고 지정합니다.
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# 모델을 '학습(fit)'시킵니다.
# - 'padded_x': 모델이 학습할 입력 데이터 (패딩된 숫자 시퀀스).
# - 'classes': 각 입력 데이터의 정답 (긍정/부정 레이블).
# - 'epochs=20': 전체 데이터셋을 20번 반복해서 학습시키라는 의미입니다.
print("\n모델 학습 시작...")
model.fit(padded_x, classes, epochs=20, verbose=0) # verbose=0으로 설정하여 학습 진행 과정을 간략하게 출력합니다.

# 학습이 끝난 모델의 성능을 평가합니다.
# 'evaluate' 함수는 테스트 데이터(여기서는 학습 데이터와 동일)에 대한 손실 값과 정확도를 반환합니다.
_, accuracy = model.evaluate(padded_x, classes, verbose=0)
print(f"\n학습 데이터에 대한 최종 정확도: {accuracy:.4f}")
````

# 마찬가지로 `y_test`도 원-핫 인코딩 형태로 변환합니다.

이 셀의 나머지 코드에 대해 자세히 설명해 드릴게요.

### 모델 컴파일: `model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])`

이 부분은 모델을 학습시키기 전에 '어떻게 학습할지' 계획을 세우는 단계와 같습니다.

*   **`loss='categorical_crossentropy'`**: 이 모델의 '손실 함수(Loss Function)'를 `categorical_crossentropy`로 설정했습니다.
    *   **손실 함수**는 모델의 예측(예: 이 뉴스는 '정치'일 확률 0.8, '경제'일 확률 0.1)이 실제 정답(예: 이 뉴스는 확실히 '정치'임)과 얼마나 **다른지 측정**하는 도구입니다. 이 값이 낮을수록 모델이 정답을 잘 맞춘다는 뜻이에요.
    *   `categorical_crossentropy`는 여러 개의 카테고리(여기서는 46개의 뉴스 주제) 중 하나를 예측하는 문제에 사용되는 아주 일반적이고 효과적인 손실 함수입니다.

*   **`optimizer='adam'`**: '최적화 도구(Optimizer)'를 `adam`으로 설정했습니다.
    *   **최적화 도구**는 모델이 학습하는 과정에서 예측이 틀렸을 때, 모델 내부의 수많은 가중치(파라미터)들을 **어떤 방향으로, 얼마나 수정해야 다음에는 더 잘 맞출지** 결정하는 역할을 합니다. `Adam`은 현재 가장 많이 사용되고 성능이 좋은 최적화 도구 중 하나입니다.

*   **`metrics=['accuracy']`**: 모델의 성능을 측정할 '평가 지표(Metric)'로 `accuracy` (정확도)를 설정했습니다.
    *   **평가 지표**는 모델이 얼마나 잘 작동하는지 **사람이 알아보기 쉬운 숫자**로 보여주는 값입니다. `accuracy`는 전체 예측 중에서 정답을 맞춘 비율을 의미해요. 예를 들어, 0.75의 정확도라면 100개의 뉴스 중 75개를 정확히 분류했다는 뜻입니다.

**간단히 말해:**
`model.compile(...)`은 "이 모델은 뉴스 주제 46개 중 하나를 예측하는 문제인데, `categorical_crossentropy`로 오차를 계산하고, `Adam` 방식으로 오차를 줄여나가며, `정확도`가 얼마나 올라가는지 보면서 학습해라"고 지시하는 것입니다.

### 조기 중단(Early Stopping) 설정: `early_stopping_callback = EarlyStopping(monitor='val_loss', patience=5)`

*   **`EarlyStopping`**: 모델이 너무 오랫동안 학습하면, 학습 데이터에만 너무 잘 맞아 떨어지고(과적합), 실제 새로운 데이터에는 오히려 성능이 나빠질 수 있습니다. `EarlyStopping`은 이런 과적합을 막기 위한 '똑똑한 정지 규칙'입니다.

    *   **`monitor='val_loss'`**: '무엇을 지켜볼지'를 설정합니다. 여기서는 `val_loss` (검증 데이터셋의 손실 값)를 지켜봅니다. 모델은 학습 데이터 외에 따로 분리해둔 '검증 데이터'를 통해 실제 성능이 어떤지 확인하는데, 이 검증 데이터에서의 오차가 줄어들어야 모델이 잘 학습되고 있다고 판단합니다.

    *   **`patience=5`**: '얼마나 기다려줄지'를 설정합니다. `val_loss`가 5번 연속으로 더 이상 좋아지지 않으면 (즉, 오차가 줄어들지 않으면) 학습을 멈추겠다는 의미입니다. 이렇게 하면 불필요한 학습을 줄이고 과적합을 방지할 수 있습니다.

**간단히 말해:**
`early_stopping_callback`은 "만약 검증 데이터의 오차(val_loss)가 5번 연속으로 나아지지 않으면, 모델이 과적합되기 시작했다고 보고 학습을 멈춰라"는 규칙을 만드는 것입니다.

### 정답 라벨 원-핫 인코딩 재변환: `y_train_one_hot = to_categorical(y_train, num_classes=46)` / `y_test_one_hot = to_categorical(y_test, num_classes=46)`

이 부분은 모델 학습 전에 `y_train`과 `y_test` (뉴스 주제 정답 라벨)의 형식을 `categorical_crossentropy` 손실 함수에 맞게 다시 한번 '원-핫 인코딩' 형태로 확실하게 변환하는 코드입니다. 이전 셀에서 이미 한 번 변환했지만, 혹시 모를 상황에 대비하거나 명확하게 하기 위해 다시 수행하는 경우가 있습니다.

*   **`to_categorical(y_train, num_classes=46)`**: `y_train`에 있는 각 뉴스 기사의 주제 번호(예: 3번, 10번 등)를 46개의 0과 1로 이루어진 벡터로 바꿔줍니다. 예를 들어, 3번 주제는 `[0, 0, 0, 1, 0, ..., 0]` (세 번째 위치만 1)처럼 됩니다.

**간단히 말해:**
이전 단계에서 `y_train`과 `y_test`를 이미 원-핫 인코딩으로 만들었지만, 모델 학습에 사용되는 데이터가 정확히 원하는 형식인지 다시 한번 확인하고 준비하는 과정입니다.

### 모델 학습 실행: `model.fit(X_train, y_train_one_hot, batch_size=100, epochs=20, validation_data=(X_test, y_test_one_hot), callbacks=[early_stopping_callback])`

이 부분이 실제로 모델을 학습시키는 핵심 코드입니다.

*   **`X_train`**: 학습에 사용할 뉴스 기사 내용 데이터입니다. (패딩된 단어 인덱스 시퀀스)

*   **`y_train_one_hot`**: `X_train`에 해당하는 뉴스 기사들의 실제 주제 정답 라벨입니다. (원-핫 인코딩된 형태)

*   **`batch_size=100`**: 학습 데이터를 한 번에 몇 개씩 모델에 넣어 학습시킬지를 정합니다. 여기서는 100개의 뉴스 기사를 한 묶음(배치)으로 묶어서 학습시킵니다. 배치 크기를 조절하면 학습 속도와 안정성에 영향을 줄 수 있습니다.

*   **`epochs=20`**: 전체 학습 데이터를 총 20번 반복해서 학습시키겠다는 의미입니다. 모델이 데이터를 여러 번 볼수록 더 잘 학습할 기회가 많아지지만, 너무 많이 보면 과적합될 수 있습니다.

*   **`validation_data=(X_test, y_test_one_hot)`**: 학습 중에 모델의 성능이 실제로는 어떤지 검증하기 위한 데이터셋을 지정합니다. 모델은 이 데이터를 '학습'하지 않고, 단순히 '평가'만 합니다. 이를 통해 모델이 새로운 데이터에 얼마나 잘 일반화되는지(`val_loss`, `val_accuracy`) 알 수 있습니다.

*   **`callbacks=[early_stopping_callback]`**: 위에서 설정한 `early_stopping_callback` 규칙을 학습 과정에 적용합니다. 검증 데이터의 손실 값이 5번 연속 개선되지 않으면 자동으로 학습을 멈추게 됩니다.

**간단히 말해:**
`model.fit(...)`은 "`X_train` 뉴스 기사들과 `y_train_one_hot` 정답 라벨을 사용해서, 100개씩 묶어 20번 반복 학습해라. 학습 중에는 `X_test`와 `y_test_one_hot`로 모델의 실제 성능을 확인하고, 만약 검증 성능이 5번 연속 안 좋아지면 학습을 중단해라"는 최종 지시입니다.

이어서 모델의 다른 층들에 대해서도 설명해 드릴게요.

### LSTM 레이어: `model.add(LSTM(100, activation='tanh'))`

**설명:**

*   **`LSTM` (Long Short-Term Memory) 레이어**: '장단기 기억'이라는 뜻을 가진 특별한 종류의 신경망 층입니다. 텍스트처럼 **순서가 중요한 데이터**를 다룰 때 아주 유용해요. 일반 신경망은 바로 앞 정보만 기억하지만, LSTM은 훨씬 더 긴 시간 동안의 정보를 기억하고 활용할 수 있습니다. 예를 들어, 문장 전체의 흐름이나 중요한 단어의 의미를 '기억'해서 다음 단어나 문장의 의미를 예측하는 데 도움을 줍니다.

    *   **`100` (첫 번째 숫자)**: 이 숫자는 LSTM 층의 '기억력' 또는 '복잡성'을 나타내는 뉴런(또는 유닛)의 개수입니다. 숫자가 클수록 모델이 더 많은 정보를 기억하고 복잡한 패턴을 학습할 수 있습니다. (임베딩 층에서 나온 각 단어의 100개 숫자 벡터를 받아 처리합니다.)

    *   **`activation='tanh'`**: 이 층에서 사용되는 '활성화 함수'입니다. 'tanh'는 '하이퍼볼릭 탄젠트'라는 수학 함수인데, LSTM 층 안에서 정보를 얼마나 보낼지, 어떻게 조절할지를 결정하는 역할을 합니다. 복잡한 계산을 통해 텍스트의 숨겨진 패턴을 찾아내는 데 기여합니다.

**간단히 말해:**
`LSTM(100, activation='tanh')`은 "단어들의 순서와 긴 문맥 정보를 100개의 뉴런으로 기억하고 처리하여 텍스트의 복잡한 패턴을 학습하는 층을 추가하라"는 의미입니다. 뉴스 기사의 흐름을 이해하는 데 중요한 역할을 합니다.

### Dense 레이어 (출력층): `model.add(Dense(46, activation='softmax'))`

**설명:**

*   **`Dense` 레이어**: 일반적인 '완전 연결' 신경망 층입니다. 이전 층(여기서는 LSTM 층)에서 처리된 모든 정보를 받아서 최종 결정을 내리는 역할을 합니다.

    *   **`46` (첫 번째 숫자)**: 이 숫자는 이 층의 뉴런(출력) 개수를 의미합니다. 이전에 `category = np.max(y_train) + 1`를 통해 총 카테고리 개수가 46개라는 것을 확인했죠? 이 모델은 46개의 뉴스 카테고리 중 하나를 예측해야 하므로, 최종적으로 46개의 출력값을 만들어냅니다.

    *   **`activation='softmax'`**: 이 층에서 사용되는 '활성화 함수'입니다. 'softmax'(소프트맥스) 함수는 46개의 출력값을 모두 합하면 1이 되도록 바꿔주는 역할을 합니다. 이렇게 하면 각 출력값이 특정 카테고리일 '확률'로 해석될 수 있습니다. 예를 들어, 첫 번째 카테고리가 0.8, 두 번째가 0.1, 세 번째가 0.05... 와 같이 모든 확률의 합이 1이 됩니다. 모델은 이 중에서 가장 높은 확률을 가진 카테고리를 최종 예측으로 선택하게 됩니다.

**간단히 말해:**
`Dense(46, activation='softmax')`은 "이전 층에서 학습한 정보를 바탕으로, 총 46개의 뉴스 카테고리 중 각각에 대해 확률을 계산하고, 가장 높은 확률을 가진 카테고리를 이 뉴스 기사의 주제로 예측하는 최종 출력층을 만들어라"는 의미입니다.

### 모델의 '입력'과 '출력' 종합

*   **입력 (Input)**: 모델의 가장 처음으로 들어가는 데이터입니다. 이 모델에서는 로이터 뉴스 기사의 **각 단어들이 숫자로 변환된 시퀀스** (그리고 `pad_sequences`로 길이가 100으로 맞춰진)가 입력이 됩니다. `Embedding` 레이어가 이 숫자 시퀀스를 받아들여 의미 있는 숫자 벡터로 바꿔주는 역할을 합니다.

*   **출력 (Output)**: 모델이 모든 계산을 마친 후 내놓는 최종 결과입니다. 이 모델에서는 **46개의 뉴스 카테고리 각각에 대한 확률 값**이 출력됩니다. 이 확률 값들 중에서 가장 높은 값을 가진 카테고리가 모델이 예측한 뉴스 기사의 최종 주제가 됩니다.

    *   **`100` (두 번째 숫자)**: 이것은 각 단어를 표현할 **숫자 벡터의 길이(차원)**를 의미합니다. 예를 들어 '경제'라는 단어가 `[0.1, -0.5, 0.2, ..., 0.8]`처럼 100개의 숫자로 이루어진 벡터로 변환된다는 뜻입니다. 이 벡터는 단어의 의미를 담고 있어서, 비슷한 의미의 단어들은 비슷한 벡터 값을 가지도록 모델이 스스로 학습합니다. 벡터의 길이가 길수록 단어의 의미를 더 세밀하게 표현할 수 있습니다.
