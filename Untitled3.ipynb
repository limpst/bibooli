# K-Fold Cross-Validation for Keras Sequential Model

This notebook demonstrates how to perform K-fold cross-validation on a Keras Sequential model. We will use the `sonar3.csv` dataset, which is a binary classification problem.

```python
# 1. Import necessary libraries
import os
import pandas as pd
from sklearn.model_selection import KFold
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.callbacks import ModelCheckpoint

print("Libraries imported successfully.")
```

## 2. Load and Prepare Data

We will load the `sonar3.csv` dataset. This dataset contains 60 features and a binary target variable (rock or mine).

```python
# Load the dataset
df = pd.read_csv("https://raw.githubusercontent.com/SoongMoo/data/refs/heads/main/sonar3.csv")

# Separate features (X) and target (y)
# X contains the first 60 columns (features)
X = df.iloc[:, 0:60]
# y contains the last column (target variable)
y = df.iloc[:, 60]

print(f"Dataset loaded. Shape: {df.shape}")
print("First 5 rows of features (X):\n", X.head())
print("First 5 rows of target (y):\n", y.head())
```

## 3. K-Fold Cross-Validation Setup

Define the number of folds and initialize the `KFold` object. We will use `shuffle=True` to randomize the data split for each fold.

```python
# Define the number of folds
k = 5

# Initialize KFold for splitting data into k folds
kfold = KFold(n_splits=k, shuffle=True, random_state=42) # Added random_state for reproducibility

# List to store accuracy scores from each fold
acc_score = []

print(f"{k}-fold cross-validation setup complete.")
```

## 4. Define the Keras Model

We define a function `model_fn` that creates and returns our Keras Sequential model. This ensures a fresh model is built for each fold.

The model consists of an `Input` layer, two `Dense` hidden layers with 'relu' activation, and a final `Dense` output layer with 'sigmoid' activation for binary classification.

```python
def model_fn():
  model = Sequential()
  # Add an Input layer as the first layer to explicitly define the input shape
  # The sonar dataset has 60 features.
  model.add(Input(shape=(60,)))
  # First Dense hidden layer
  model.add(Dense(24, activation='relu'))
  # Second Dense hidden layer
  model.add(Dense(10, activation='relu'))
  # Output layer for binary classification (0 or 1)
  model.add(Dense(1, activation='sigmoid'))
  return model

print("Model function 'model_fn' defined.")
```

## 5. Perform K-Fold Cross-Validation

This loop iterates through each fold, splitting the data into training and test sets, building and compiling a new model, training it, and evaluating its performance. The best model from each fold (based on validation loss) is saved.

```python
# Prepare directory for saving models from each fold
model_save_dir = 'best_models'
os.makedirs(model_save_dir, exist_ok=True)

print(f"Models will be saved to: {model_save_dir}")

# Perform K-fold cross-validation
for fold, (train_index, test_index) in enumerate(kfold.split(X)):
  print(f"\n--- Training Fold {fold+1}/{k} ---")

  # Split data for the current fold
  X_train, X_test = X.iloc[train_index, :], X.iloc[test_index, :]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]

  # Create a new model for each fold to ensure independent training
  model = model_fn()

  # Compile the model for binary classification
  model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

  # Define ModelCheckpoint to save the best model from this fold
  # The model is saved if 'val_loss' improves. The filename will include the fold number and validation accuracy.
  filepath = os.path.join(model_save_dir, f"model_fold_{fold+1}_val_acc_{{val_accuracy:.4f}}.keras")
  checkpointer = ModelCheckpoint(filepath=filepath, monitor='val_loss', verbose=0, save_best_only=True)

  # Train the model with a validation split and callbacks
  # validation_split carves out a portion of the training data for validation during training
  # verbose=0 suppresses training output for cleaner logs.
  history = model.fit(X_train, y_train, epochs=50, batch_size=500,
                      validation_split=0.25, callbacks=[checkpointer], verbose=0)

  # Evaluate the model on the test set for this fold
  _, accuracy = model.evaluate(X_test, y_test, verbose=0)
  acc_score.append(accuracy)
  print(f"Fold {fold+1} Test Accuracy: {accuracy:.4f}")

print("\nK-fold cross-validation complete.")
```

## 6. Summarize Results

Calculate and print the average accuracy across all folds to get an overall performance estimate of the model.

```python
# Calculate and print the average accuracy across all folds
avg_acc_score = sum(acc_score) / k
print(f"\nAccuracy scores for each fold: {acc_score}")
print(f"Average K-fold Cross-Validation Accuracy: {avg_acc_score:.4f}")
```
